# 例1.1（ガウスの消去法で3元連立一次方程式を解く）
$$
x_1 + x_2 + 3x_3 = 12\\
x_1 + 3x_2 + 2x_3 =13\\
2x_1 + 6x_2 + x_3 =17
$$
を行列で書くと、**Ax=b**
$$
\mathbf{A}=\begin{pmatrix}
   1 & 1 & 3\\
   1 & 3 & 2\\
   2 & 6 & 1
\end{pmatrix},
\mathbf{x}=\begin{pmatrix}
   x_1\\
   x_2\\
   x_3
\end{pmatrix},
\mathbf{b}=\begin{pmatrix}
   12\\
   13\\
   17
\end{pmatrix},
$$

```python
A = [[1, 1, 3],
     [1, 3, 2],
     [2, 6, 1]]
b = [12, 13, 17]
```

```python
# 行列Aとベクトルbを結合
augmented_matrix = [A[i] + [b[i]] for i in range(3)] #3x4行列
print(augmented_matrix)
```

```python
n=3 #次元数
#前進消去
for i in range(n):
  pivot = augmented_matrix[i][i] #行の最も左にあるゼロでない成分
  print(pivot)
  for j in range(i + 1, n):
    factor = augmented_matrix[j][i] / pivot
    for k in range(n+1):
      augmented_matrix[j][k] = augmented_matrix[j][k] - factor * augmented_matrix[i][k]
  # ピボットで割る
  for k in range(i, n + 1):
    augmented_matrix[i][k] = augmented_matrix[i][k]/pivot
print(augmented_matrix)
```

```python
#後退代入
x = [0,0,0]
for i in range(n-1,-1,-1): #最後の-1に注目。1ずつ後退という指示
  x[i] = augmented_matrix[i][n]
  for j in range(i + 1, n):
    x[i] = x[i] - augmented_matrix[i][j] * x[j]
print(x)
```

# 例1.2（ガウスの消去法の計算量）
適当な100元連立方程式を設定して、その方程式を解くのにかかる時間を計測する。

```python
# prompt: 100元連立方程式を設定して、その方程式を解くのにかかる時間を計測するコード

import random
import time

def gaussian_elimination(A, b):
    n = len(A)
    augmented_matrix = [A[i] + [b[i]] for i in range(n)]

    # 前進消去
    for i in range(n):
        # ピボット選択（部分ピボット選択）
        max_row = i
        for k in range(i + 1, n):
            if abs(augmented_matrix[k][i]) > abs(augmented_matrix[max_row][i]):
                max_row = k
        augmented_matrix[i], augmented_matrix[max_row] = augmented_matrix[max_row], augmented_matrix[i]

        pivot = augmented_matrix[i][i]
        if abs(pivot) < 1e-10:  # ピボットがゼロに近い場合の処理
            raise ZeroDivisionError("Pivot is zero or near zero.")

        for j in range(i + 1, n):
            factor = augmented_matrix[j][i] / pivot
            for k in range(i, n + 1):
                augmented_matrix[j][k] -= factor * augmented_matrix[i][k]

    # 後退代入
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = augmented_matrix[i][n]
        for j in range(i + 1, n):
            x[i] -= augmented_matrix[i][j] * x[j]
        x[i] /= augmented_matrix[i][i]  # ピボットで割る

    return x
# def gaussian_eliminationはここまで ============================

# 100元連立方程式をランダムに生成
n = 100
A = [[random.random() for _ in range(n)] for _ in range(n)]
b = [random.random() for _ in range(n)]

# 時間計測
start_time = time.time()
try:
    x = gaussian_elimination(A, b)
    # print(x)  # 解を表示しない
except ZeroDivisionError as e:
    print(e)
except Exception as e:
    print(f"An error occurred: {e}")
end_time = time.time()

elapsed_time = end_time - start_time
print(f"計算時間: {elapsed_time} 秒")
```

# LU分解
適当な100元連立一次方程式をLU分解を用いて解く計算にかかる時間を計測する。

```python
# prompt: 100元連立方程式を設定して、その方程式をLU分解を用いて解くのにかかる時間を計測するコード。numpyは使わない。

import random
import time

def gaussian_elimination(A, b):
    n = len(A)
    augmented_matrix = [A[i] + [b[i]] for i in range(n)]

    # 前進消去
    for i in range(n):
        # ピボット選択（部分ピボット選択）
        max_row = i
        for k in range(i + 1, n):
            if abs(augmented_matrix[k][i]) > abs(augmented_matrix[max_row][i]):
                max_row = k
        augmented_matrix[i], augmented_matrix[max_row] = augmented_matrix[max_row], augmented_matrix[i]

        pivot = augmented_matrix[i][i]
        if abs(pivot) < 1e-10:  # ピボットがゼロに近い場合の処理
            raise ZeroDivisionError("Pivot is zero or near zero.")

        for j in range(i + 1, n):
            factor = augmented_matrix[j][i] / pivot
            for k in range(i, n + 1):
                augmented_matrix[j][k] -= factor * augmented_matrix[i][k]

    # 後退代入
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = augmented_matrix[i][n]
        for j in range(i + 1, n):
            x[i] -= augmented_matrix[i][j] * x[j]
        x[i] /= augmented_matrix[i][i]  # ピボットで割る

    return x

def lu_decomposition(A):
    n = len(A)
    L = [[0.0] * n for _ in range(n)]
    U = [[0.0] * n for _ in range(n)]

    for i in range(n):
        L[i][i] = 1.0
        for j in range(i, n):
            sum_u = sum(L[i][k] * U[k][j] for k in range(i))
            U[i][j] = A[i][j] - sum_u
        for j in range(i + 1, n):
            sum_l = sum(L[j][k] * U[k][i] for k in range(i))
            L[j][i] = (A[j][i] - sum_l) / U[i][i]
    return L, U

def solve_lu(L, U, b):
    n = len(L)
    y = [0.0] * n
    x = [0.0] * n

    # Ly = b
    for i in range(n):
        sum_y = sum(L[i][k] * y[k] for k in range(i))
        y[i] = b[i] - sum_y

    # Ux = y
    for i in range(n-1, -1, -1):
        sum_x = sum(U[i][k] * x[k] for k in range(i+1, n))
        x[i] = (y[i] - sum_x) / U[i][i]

    return x


# 100元連立方程式をランダムに生成
n = 100
A = [[random.random() for _ in range(n)] for _ in range(n)]
b = [random.random() for _ in range(n)]

# LU分解を用いた解法時間計測
start_time = time.time()
try:
    L, U = lu_decomposition(A)
    x = solve_lu(L, U, b)
    #print(x)
except ZeroDivisionError as e:
    print(e)
except Exception as e:
    print(f"An error occurred: {e}")
end_time = time.time()
elapsed_time_lu = end_time - start_time
print(f"LU分解計算時間: {elapsed_time_lu} 秒")
```
