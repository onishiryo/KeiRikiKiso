# 例1.1（ガウスの消去法で3元連立一次方程式を解く）
$$
x_1 + x_2 + 3x_3 = 12\\
x_1 + 3x_2 + 2x_3 =13\\
2x_1 + 6x_2 + x_3 =17\\

\mathbf{A}=\begin{pmatrix}
   1 & 1 & 3\\
   1 & 3 & 2\\
   2 & 6 & 1
\end{pmatrix}, \\
$$

```python
# prompt: ガウスの消去法で三元連立一次方程式を解くコードを作成して。numpyは使わない。

def gaussian_elimination(A, b):
    """
    ガウスの消去法で三元連立一次方程式を解く。

    Args:
        A: 係数行列 (3x3 list of lists)
        b: 定数項ベクトル (3x1 list)

    Returns:
        解ベクトル (3x1 list) または None (解が存在しない場合)
    """

    n = len(A)
    if n != 3 or len(A[0]) != 3 or len(b) != 3:
        print("入力行列のサイズが不正です。3x3の係数行列と3x1の定数項ベクトルを入力してください。")
        return None

    # 行列Aとベクトルbを結合
    augmented_matrix = [A[i] + [b[i]] for i in range(n)]

    #前進消去
    for i in range(n):
        # ピボットを探す
        pivot = augmented_matrix[i][i]
        if pivot == 0:
          # ピボットがゼロの場合、次の行と入れ替える
          for j in range(i + 1, n):
            if augmented_matrix[j][i] != 0:
              augmented_matrix[i], augmented_matrix[j] = augmented_matrix[j], augmented_matrix[i]
              pivot = augmented_matrix[i][i]
              break

          if pivot == 0:  # ピボットがゼロのままなら解が存在しない可能性がある
            print("解が存在しない可能性があります。")
            return None


        # ピボットで割る
        for k in range(i, n + 1):
            augmented_matrix[i][k] /= pivot

        # 他の行から引き算する
        for j in range(n):
            if j != i:
                factor = augmented_matrix[j][i]
                for k in range(i, n + 1):
                    augmented_matrix[j][k] -= factor * augmented_matrix[i][k]

    # 後退代入
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = augmented_matrix[i][n]
        for j in range(i + 1, n):
            x[i] -= augmented_matrix[i][j] * x[j]

    return x

# 使用例
A = [[2, 1, -1],
     [-3, -1, 2],
     [-2, 1, 2]]
b = [-1, 4, -5]

solution = gaussian_elimination(A, b)

if solution:
    print("解:", solution)
```
