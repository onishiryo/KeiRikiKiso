***行列・ベクトル計算***

numpyライブラリを使わずに、行列・ベクトル計算をやってみよう。
numpyライブラリを利用するともっと簡単に行列・ベクトル計算をできます。
その方法については後ほど説明します。

a=[1,2,3], b=[-1,0,1]の内積を計算

````python
vec_a=[ 1.0, 2.0, 3.0] #vector a
vec_b=[-1.0, 0.0, 1.0] #vector b
````

````python
inpro = 0.0 #inner product
for i in [0,1,2]:
  inpro=inpro + vec_a[i]*vec_b[i]
print('inner product=', inpro)
````

for文による繰り返しの範囲について：
rangeを使うと便利です。range(X)は0からX-1までのX個の整数リストです

````python
inpro = 0.0 #inner product
for i in range(3):  #range(3)=[0,1,2]
  inpro=inpro + vec_a[i]*vec_b[i]
print('inner product=', inpro)
````

行列演算

```
A=[0,1]
  [1,2]
B=[-2,-1]
  [-1,0]
```
の積を計算してみましょう。

````python
A=([0.0, 1.0],[1.0, 2.0])
B=([-2.0, -1.0],[-1.0, 0.0])
````

````python
行列をprintするとこんな感じに出力されます↓
````

````python
print(A)
````

````python
AB_11=A[0][0]*B[0][0]+A[0][1]*B[1][0]
AB_12=A[0][0]*B[0][1]+A[0][1]*B[1][1]
AB_21=A[1][0]*B[0][0]+A[1][1]*B[1][0]
AB_22=A[1][0]*B[0][1]+A[1][1]*B[1][1]
````

````python
print('AB=|',AB_11, AB_12, '|')
print('   |',AB_21, AB_22, '|')
````

これでは、要素数や次元が大きくなったときに大変。
次は、for文を使って書いてみましょう。

まずは、A*Bの要素を格納する2次元（2x2）配列ABを用意します。
すべての要素がゼロである配列ができます。

````python
AB = [[0]*2]*2
print(AB)
````

あれprintされたABがおかしい！

````python
for i in range(2): #i=0,1
  for j in range(2): #j=0,1
    AB[i][j]=A[i][0]*B[0][j]+A[i][1]*B[1][j]
    print('AB[',i,'][',j,']=', AB[i][j])
````

````python
print(AB)
print(AB[0][1])
print(AB[1][0])
````
