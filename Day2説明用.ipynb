# **ベクトルとベクトルの乗算**

# 例1.1
a=[1,2,3], b=[-1,0,1]の内積を計算

````python
vec_a=[ 1, 2, 3] #vector a
vec_b=[-1, 0, 1] #vector b
````

````python
inpro=vec_a[0]*vec_b[0]+vec_a[1]*vec_b[1]+vec_a[2]*vec_b[2]
print(inpro)
````

forループ（繰り返し）を使ってみよう

````python
inpro = 0.0 #inner product
for i in [0,1,2]:
  inpro=inpro + vec_a[i]*vec_b[i]
print('inner product=', inpro)
````

for文による繰り返しの範囲について：
rangeを使うと便利です。range(X)は0からX-1までのX個の整数リストです

````python
inpro = 0.0 #inner product
for i in range(3):  #range(3)=[0,1,2]
  inpro=inpro + vec_a[i]*vec_b[i]
print('inner product=', inpro)
````

# ベクトルと行列の乗算

# **例1.2**

```
2x3行列A=|0,1,2|
         |1,4,9|
```
と
```
ベクトルb=(-1,0,1)
```
の積を求める


````python
#2x3行列A=|0,1,2|
#         |1,4,9|
A=[[0,1,2], [1,4,9]]
#ベクトルb
vec_b=[-1,0,1]
````

行列をprintすると、どう表示されるか見てみる。

````python
print(A)
````

````python
Ab_1=A[0][0]*b[0]+A[0][1]*b[1]+A[0][2]*b[2]
Ab_2=A[1][0]*b[0]+A[1][1]*b[1]+A[1][2]*b[2]
````

````python
print(Ab_1)
print(Ab_2)
````

forループを使って計算してみる

````python
#答え用の行列Abを用意しておく
#答えは2x1行列になる。しかし、ここでは1x2行列を用意した。詳細は【補足】で説明する。
Ab=[0,0]
````

# 例1.4
次は2x4行列Aと4x2行列Bの積を計算します。

```
A=|1,-1,0,4|
  |2, 0,1,2|
B=|1, 2|
  |2,-1|
  |3, 0|
  |4,-3|
AB=|AB_11, AB_12|
   |AB_21, AB_22|
```



````python
#ex1.4 A&B
A=[[1,-1,0,4],[2,0,1,2]]
B=[[1,2],[2,-1],[3,0],[4,-3]]
````

````python
AB_11=A[0][0]*B[0][0]+A[0][1]*B[1][0]+A[0][2]*B[2][0]+A[0][3]*B[3][0]
AB_12=A[0][0]*B[0][1]+A[0][1]*B[1][1]+A[0][2]*B[2][1]+A[0][3]*B[3][1]
AB_21=A[1][0]*B[0][0]+A[1][1]*B[1][0]+A[1][2]*B[2][0]+A[1][3]*B[3][0]
AB_22=A[1][0]*B[0][1]+A[1][1]*B[1][1]+A[1][2]*B[2][1]+A[1][3]*B[3][1]
````

````python
print('AB=|',AB_11, AB_12, '|')
print('   |',AB_21, AB_22, '|')
````

次は、for文を使って書いてみましょう。

まずは、A*Bの要素を格納する2次元（2x2）配列ABを用意します。
すべての要素がゼロである配列とします。
まずは例1.4の場合から始めます。ex1.3 A&B と書かれたコードボックスを実行することで、例1.4の行列AとBをロードしてください。（colab練習のために、あえてややこしいやり方をしています。）

````python
#解を格納する2x2変数を用意。初期はすべての要素をゼロとする。
AB = ([0,0],[0,0])
print(AB)
````

````python
for i in range(2): #i=0,1
  for j in range(2): #j=0,1
    AB[i][j]=A[i][0]*B[0][j]+A[i][1]*B[1][j]
    print('AB[',i,'][',j,']=', AB[i][j])
````

````python
print('AB=|',AB[0][0], AB[0][1], '|')
print('   |',AB[1][0], AB[1][1], '|')
````

次に、例1.4の計算をforループを使って計算しよう。

````python
A=[[1,-1,0,4],[2,0,1,2]]
B=[[1,2],[2,-1],[3,0],[4,-3]]
````

````python
#再び、解を格納する2x2変数を用意。初期はすべての要素をゼロとする。（3つ上のコードボックスを実行するのと同意）
AB = ([0,0],[0,0])
print(AB)
````

````python
for i in range(2): #i=0,1
  for j in range(2): #j=0,1
    AB[i][j]=A[i][0]*B[0][j]+A[i][1]*B[1][j]+A[i][2]*B[2][j]+A[i][3]*B[3][j]
    print('AB[',i,'][',j,']=', AB[i][j])
````

````python
print('AB=|',AB[0][0], AB[0][1], '|')
print('   |',AB[1][0], AB[1][1], '|')
````

一行がかなり長くなった。もう一つforループを使った方が良いんじゃないか！？

````python
AB=[[0]*2 for i in range(2)] #すべての要素をゼロ初期化
#↓この形が行列の乗算の基本
for i in range(2): #i=0,1
  for j in range(2): #j=0,1
    for k in range(4): #k=0,1,2,3
      AB[i][j]=AB[i][j] +A[i][k]*B[k][j]
    print('AB[',i,'][',j,']=', AB[i][j])
````

総和規約を使うとわかりやすい。

$$(AB)_{ij}=\Sigma_k{a_{ik}b_{kj}}=a_{ik}b_{kj}$$

ここで、下付きは要素番号を表している。
- $(AB)_{ij}$は行列ABの$(i,j)$要素、
- $a_{ik}$は行列Aの$(i,k)$要素、
- $b_{kj}$は行列Bの$(k,j)$要素である。

Aが2x4、Bが4x2行列だったことから、i,k,jはそれぞれ2,4,2次元であることがわかる。だから、for文中のrangeがそれぞれ2,4,2であった。

# （補足）ベクトルについて
n次元ベクトルは、nx1行列とも考えられる（横にn個数字が並ぶのではなく、縦にn個数字が並ぶ）。
2x4行列と4次元ベクトルの積は2x4行列と4x1行列の積と考えることもできる。そうすると、行列同士の乗算と同じである。

以下は、試しにAIにコードを生成してもらった。

````python
# prompt: 2x4行列と4x1行列の積を求めるコード

# 2x4行列A
A = [[1, 2, 3, 4],
     [5, 6, 7, 8]]

# 4x1行列B
B = [[9],
     [10],
     [11],
     [12]]

# 結果を格納する2x1行列Cを初期化
C = [[0], [0]]

# 行列の積を計算
for i in range(2):  # 行
    for j in range(1):  # 列
        for k in range(4):  # kは和を取るためのループ変数
            C[i][j] += A[i][k] * B[k][j]

# 結果を表示
C
````
