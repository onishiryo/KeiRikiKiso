```python
# ファイルパス
_filepath = './test.cpp'

# ファイルに書き込む内容
filecontents = """
#include <stdio.h>
#include <float.h>

int main(void){
	float a=1.0;
	float b=1.0/3.0;
	if(a==b*3.0){
		printf("a is equal to b*3.0");
	}
  else{
		printf("a is not equal to b*3.0");
    printf("  a=%f", a);
    printf("  b*3.0=%f", b*3.0);
	}
}
"""

# 書き込みモード
with open(_filepath, 'w') as f:
  # ファイル作成、書き込み
  f.write(filecontents)
```
