```python
# prompt: 最急降下法を使って3元連立方程式を解く。係数行列を導入すること。勾配ベクトル、残差ベクトル、修正係数を計算すること。日本語での説明をつける。numpyを使ってはいけない。
# 注意：上記プロンプトでは正解に至らなかった。

import math

def SDM_3d(A, b):
    """
    最急降下法を使って3元連立方程式を解く。

    Args:
        A: 係数行列 (3x3 リスト)
        b: 定数項ベクトル (長さ3のリスト)

    Returns:
        解ベクトル (長さ3のリスト) または None (収束しなかった場合)
    """


    # 初期設定
    tolerance=1e-4
    max_iterations=3

    # 初期推定値
    initial_guess = [0, 0, 0]
    x = initial_guess[:]

    # 初期残差ベクトルの計算
    residual = [0, 0, 0]
    residual[0] = b[0] - (A[0][0] * x[0] + A[0][1] * x[1] + A[0][2] * x[2])
    residual[1] = b[1] - (A[1][0] * x[0] + A[1][1] * x[1] + A[1][2] * x[2])
    residual[2] = b[2] - (A[2][0] * x[0] + A[2][1] * x[1] + A[2][2] * x[2])
    print("residual0=", residual)

    l2_b = math.sqrt(b[0]**2 + b[1]**2 + b[2]**2)

    for itr in range(max_iterations):
        # 修正係数alphaを計算。SDMでは修正方向ベクトルと残差ベクトルは同じ。
        alpha_numerator = residual[0]**2 + residual[1]**2 + residual[2]**2
        alpha_denominator = \
            residual[0] * (A[0][0] * residual[0] + A[0][1] * residual[1] + A[0][2] * residual[2]) + \
            residual[1] * (A[1][0] * residual[0] + A[1][1] * residual[1] + A[1][2] * residual[2]) + \
            residual[2] * (A[2][0] * residual[0] + A[2][1] * residual[1] + A[2][2] * residual[2])
        alpha = alpha_numerator / alpha_denominator

        # 解ベクトルの更新。SDMでは修正方向ベクトルと残差ベクトルは同じ。
        x[0] += alpha * residual[0]
        x[1] += alpha * residual[1]
        x[2] += alpha * residual[2]

        # 更新した残差ベクトルの計算
        residual[0] = b[0] - (A[0][0] * x[0] + A[0][1] * x[1] + A[0][2] * x[2])
        residual[1] = b[1] - (A[1][0] * x[0] + A[1][1] * x[1] + A[1][2] * x[2])
        residual[2] = b[2] - (A[2][0] * x[0] + A[2][1] * x[1] + A[2][2] * x[2])

        # 収束判定 (勾配のノルムが許容誤差以下)
        l2_residual = math.sqrt(residual[0]**2 + residual[1]**2 + residual[2]**2)
        relative_error = l2_residual / l2_b

        print(f"Iteration {itr+1}: Relative Error = {relative_error}")
        print("alpha=", alpha)
        print("residual=", residual)
        print(x)
        if relative_error < tolerance:
            return x

    return None  # 最大反復回数に達しても収束しなかった場合


# 使用例
A = [[2, 1, -1],
     [-3, -1, 2],
     [-2, 1, 2]]
b = [0, 1, 2]
# A = [[1, 1, 0],
#      [0, 1, 0],
#      [0, 0, 2]]
# b = [3, 1, 2]

solution = SDM_3d(A, b)

if solution:
    print("解:", solution)
else:
    print("解が見つかりませんでした。")
```
